```python
import numpy as np
import matplotlib.pyplot as plt
from pymultifit.distributions.utilities_d import q_exponential_pdf_

x = np.linspace(0, 3, 500)
q_values = [0.5, 1.0, 1.5]
rate = 1.0

plt.figure(figsize=(8, 5))

for q in q_values:
    y = q_exponential_pdf_(x, q=q, rate=rate)
    plt.plot(x, y, label=f"q = {q}")

# Visual marker for finite upper bound cutoff when q < 1 (at x = 1 / (rate * (1 - q)))
x_cutoff = 1.0 / (rate * (1.0 - 0.5))
plt.axvline(x_cutoff, color='red', linestyle='--', alpha=0.7, label=f'q=0.5 Cutoff (x={x_cutoff})')

plt.title("q-Exponential PDF: Finite Support (q < 1) vs Heavy Tail (q > 1)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.ylim(-0.05, 1.05)
plt.legend()
plt.grid(True)
plt.show()
```
